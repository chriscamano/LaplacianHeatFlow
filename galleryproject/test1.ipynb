```python
import scipy.io as sio
import numpy as np
import matplotlib.pyplot as plt
from scipy.sparse import csr_matrix
import seaborn as sns

def load_and_visualize_matrix(file_path):
    # Load the .mat file
    mat_contents = sio.loadmat(file_path)
    
    # Get the matrix - assuming it's stored with the name 'pre2'
    matrix = mat_contents['pre2']
    
    # Convert to CSR format if not already
    if not isinstance(matrix, csr_matrix):
        matrix = csr_matrix(matrix)
    
    # Print basic information
    print(f"Matrix shape: {matrix.shape}")
    print(f"Number of nonzero elements: {matrix.nnz}")
    print(f"Sparsity: {matrix.nnz / (matrix.shape[0] * matrix.shape[1]) * 100:.2f}%")
    
    # Create figure with multiple subplots
    plt.figure(figsize=(15, 5))
    
    # 1. Spy plot (structure visualization)
    plt.subplot(131)
    plt.spy(matrix, markersize=0.1)
    plt.title('Matrix Structure')
    
    # 2. Heatmap of a small section (top-left corner)
    plt.subplot(132)
    dense_corner = matrix[:50, :50].toarray()
    sns.heatmap(dense_corner, cmap='coolwarm', center=0)
    plt.title('Top-left Corner Heatmap (50x50)')
    
    # 3. Distribution of nonzero values
    plt.subplot(133)
    values = matrix.data
    plt.hist(values, bins=50, density=True)
    plt.title('Distribution of Nonzero Values')
    plt.xlabel('Value')
    plt.ylabel('Density')
    
    plt.tight_layout()
    plt.show()
    
    return matrix

# Load and visualize the matrix
file_path = '../pre2.mat'
matrix = load_and_visualize_matrix(file_path)
```
